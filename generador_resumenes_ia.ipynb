Este proyecto implementa un sistema de resumen automático de textos en español usando el algoritmo **LSA (Latent Semantic Analysis)**. El sistema extrae las oraciones más relevantes de un texto largo y las presenta como resumen.

**Librerías utilizadas:**
- `sumy` — generación de resúmenes mediante NLP
- `nltk` — tokenización del texto en español

## 📥 Paso 2 — Descarga de recursos de NLTK

NLTK necesita descargar los tokenizadores de idioma para poder segmentar el texto en oraciones correctamente.

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

## ✍️ Paso 3 — Ingreso del texto

Pega cualquier texto en español que desees resumir. Funciona mejor con textos de 3 oraciones o más.

```python
# Ingresa el texto que deseas resumir
texto = input("Pega aquí tu texto: ")

print("\n" + "="*50)
print("TEXTO ORIGINAL")
print("="*50)
print(texto)
```

## 🤖 Paso 4 — Generación del resumen con NLP

Se utiliza el algoritmo **LSA (Latent Semantic Analysis)** de la librería Sumy. Este algoritmo identifica los conceptos principales del texto mediante álgebra matricial y selecciona las oraciones más representativas.

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer

# Procesar el texto con tokenizador en español
parser = PlaintextParser.from_string(texto, Tokenizer("spanish"))

# Crear el resumidor LSA
summarizer = LsaSummarizer()

# Generar resumen con las 2 oraciones más importantes
resumen = summarizer(parser.document, sentences_count=2)

print("\n" + "="*50)
print("RESUMEN GENERADO")
print("="*50)
for frase in resumen:
    print("-", frase)
```
